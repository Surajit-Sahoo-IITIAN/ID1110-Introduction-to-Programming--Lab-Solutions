## Q.1. Checking Whether a String Is a Palindrome and Its Distance to Palindrome ##
- A string is a palindrome if  
  $S[i] = S[n-1-i]$ for all $i < \lfloor n/2 \rfloor$
- **Distance to Palindrome**:
  - For a given string \( S \) of length \( n \),
  - the distance to palindrome is the **minimum number of character changes**
    required to make \( S \) a palindrome.
  - Formally, it is the number of indices \( i \) such that  
    $( S[i] \neq S[n-1-i] ) \ for ( 0 \le i < \lfloor n/2 \rfloor )$.
- **Build-in Functions Used**:
    - **`lower()`**: Converts all uppercase characters in the string to lowercase, enabling case-insensitive comparison.
    - **`replace(" ", "")`**: Removes all spaces from the string so that only characters are considered while checking for palindrome.



```python
def Palindrome_Check(Str):
    print(Str)
    count = 0
    is_palindrome = True
    for i in range(len(Str) // 2):
        if Str[i].lower() != Str[len(Str) - 1 - i].lower():
            is_palindrome = False
            count += 1
    return is_palindrome, count
Str=input("Enter the string to check Palindrome")
is_palindrome,count=Palindrome_Check(Str.replace(" ",""))
if is_palindrome:
        print("It is a palindrome")
else:
        print("Not a palindrome")
        print("Distance to palindrome:", count)
```

## Q.2. Binary Representation Without Using Built-in Conversion Functions ##

**Algorithm:**
- Initialize `current` to `num`
- Initialize a list `l` to an empty list
- Repeatedly do the following while `current > 0`:
  - Compute `remainder` as `current % 2`
  - Append `remainder` to the list `l`
  - Update `current` as `current // 2`
- Create a string `bin_string` by reading the elements of `l` in reverse order
- Print `bin_string`


```python
def decimal_to_binary(num):
    if num <= 0:
        print("invalid input")
        return
    current = num
    l = []
    while current > 0:
        remainder = current % 2
        l.append(remainder)
        current = current // 2
    bin_string = ""
    for i in range(1,len(l)+1):
        bin_string = bin_string + str(l[len(l)-i])

    print("Binary representation:", bin_string)
num = int(input("Enter a positive integer: "))
decimal_to_binary(num)
```

## Q.3. Prime Number Generation Using a Function ##
**Algorithm:**
- Read two positive integers `m` and `k`
- If either `m` or `k` is less than or equal to zero, print "invalid input" and asking for valid input
- Initialize an empty list `primes`
- Initialize a counter variable `count` to 0
- Initialize a variable `current` to `m`
- While `count < k`, do the following:
  - Check if `current` is prime using the `isprime` function
  - If `current` is prime:
    - Append `current` to the list `primes`
    - Increment `count` by 1
  - Increment `current` by 1
- Print the list `primes`

**Debugging:**
- **Breakpoint:** A breakpoint pauses program execution at a specific line to inspect program behavior.
- **Step Into:** Executes the current line and enters into the called function line by line.
- **Step Over:** Executes the current line without entering into any called function.
- **Resume:** Continues program execution normally until the next breakpoint or program termination.

```python
def isprime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
m = int(input("Enter the value of m: "))
k = int(input("Enter the value of k: "))
while m <= 0 or k <= 0:
    print("invalid input")
    m = int(input("Enter the value of m: "))
    k = int(input("Enter the value of k: "))
else:
    primes = []
    count = 0
    current = m
    while count < k:
        if isprime(current):
            primes.append(current)
            count += 1
        current += 1
    print(f"{k} Prime numbers greater than or equal to {m}:{primes}")
```

## Q.4. Binary Conversion of List Elements Using a User-Defined Function ##
- Use the same Algorithm as in **Q.2.**

```python
def ConvertToBinary(num):
    current = num
    l = []
    while current > 0:
        remainder = current % 2
        l.append(remainder)
        current = current // 2
    bin_string = ""
    for i in range(1,len(l)+1):
        bin_string = bin_string + str(l[len(l)-i])
    return bin_string
n = int(input("Enter the size of the list: "))
l = []
for i in range(n):
    num = int(input("Enter a non-negative integer: "))
    while num < 0:
        print("Invalid input. Enter a non-negative integer.")
        num = int(input("Enter a non-negative integer: "))
    l.append(num)
print("\nNumber  Binary Representation")
for num in l:
    binary = ConvertToBinary(num)
    print(num, "     ", binary)
```

## Q.5.Student Grade Processing and Frequency Computation: ##

**Step 1: `ComputeGrade Function`**
- **Input**: An valid integer (between $[0,100]$) representing marks.  
- **Output**: A character denoting the grade (`S`, `A`, `B`, `C`, `D`, `E`, `F`) or `I` for invalid marks.


```python
def ComputeGrade(marks):
    if marks < 0 or marks > 100:
        return 'I'
    elif marks >= 90:
        return 'S'
    elif marks >= 75:
        return 'A'
    elif marks >= 65:
        return 'B'
    elif marks >= 55:
        return 'C'
    elif marks >= 45:
        return 'D'
    elif marks >= 35:
        return 'E'
    else:
        return 'F'
        
```

**Step 2: `ComputeFrequencies Function`**  
- **Input**: A list of grades and a list of possible grade keys.  
- **Output**: A list containing the number of students corresponding to each grade.


```python
def ComputeFrequencies(elist, keylist):
    freq = []
    for key in keylist:
        count = 0
        for elem in elist:
            if elem == key:
                count += 1
        freq.append(count)
    return freq
```

**Step 3: `Main Program`**  
- **Input**: Number of students and marks of each student (with validation).  
- **Output**: Prints marks with their corresponding grades and prints each grade with the number
  of students who obtained that grade.

```python
n = int(input("Enter number of students: "))
marks_list = []
grades_list = []
for i in range(1, n + 1):
    marks = int(input(f"Enter marks of student {i}: "))
    while marks < 0 or marks > 100:
        print("Invalid marks. Enter marks between 0 and 100.")
        marks = int(input(f"Enter marks of student {i}: "))
    marks_list.append(marks)
    grade = ComputeGrade(marks)
    grades_list.append(grade)

print("\nMarks and Grades:")
for i in range(n):
    print("Student", i + 1, ":", marks_list[i], "-", grades_list[i])

grade_keys = ['S', 'A', 'B', 'C', 'D', 'E', 'F']
freq = ComputeFrequencies(grades_list, grade_keys)

print("\nGrade Distribution:")
for i in range(len(grade_keys)):
    print(grade_keys[i], ":", freq[i])
```
