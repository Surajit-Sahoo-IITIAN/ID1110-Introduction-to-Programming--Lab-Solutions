## Q.1. Removing Duplicates from a list ##

- Step 1: Write a function `remove_duplicates` that takes a list **lst1** as parameter and returns
a list **lst** which is obtained by removing duplicate elements from **lst1**. It must be
ensured that for any two elements $a$ and $b$ in **lst1** with $a \neq b$, if in **lst1** the first
occurrence of $a$ is before the first occurrence of $b$, then in **lst**, $a$ appears before $b$.

```python
def remove_duplicates(lst1):
    lst=[]
    for i in lst1:
        if i not in lst:
                lst.append(i)
    return lst
```

- Step 2: Write a main program that does the following:
    - Ask the user to enter a positive integer $n$.
    - If $n \le 0$, terminate the program.
    - If $n > 0$, ask the user to enter $n$ integers and store these $n$ numbers in a list $l_1$.
    - Invoke the `remove_duplicates` function to remove duplicates from $l_1$ and store
    the result in a new list $l$.
    - Print the list $l$.
    - For every pair of elements at positions $(i, j)$ in $l$ such that $i < j$, do the
    following:
        - check whether $l[i]$ and $l[j]$ have the same remainder when they are divided by
        $7$ and if so, then print the pair of elements $(l[i], l[j])$.

```python
n=int(input("Enter a integer"))
if n<=0:
    print("Terminating the program...")
    exit 
else:
    print(f"Enter the {n} integers:")
    l1=[]
    for i in range(n):
        n=int(input(f"Enter the {i+1} th integer"))
        l1.append(n)
    l=remove_duplicates(l1)
    print(f"Modified list after removing duplicates:{l}")
    print("The required pair of elements are:")
    for i in range(len(l)):
        for j in range(i+1,len(l)):
            if l[i]%7==l[j]%7:
                print(f"({l[i]},{l[j]})", end=" ")
```

## Q.2. Recursive Function Implementation ##

Consider the sequence whose first three terms are $g(0) = 1$, $g(1) = 2$, and
$g(2) = 3$ and for $n \ge 3$ the terms satisfy the recursive relation 
$$g(n) = g(n − 1) + g(n − 2) + g(n − 3) − 1$$.
 Thus the sequence starts as 1, 2, 3, 5, 9, 16, . . ..

- Step 1: Write a recursive function `compute_terms` that takes an integer $n \ge 0$ as parameter
and returns the triple $(g(n), g(n − 1), g(n − 2))$. For convenience, you may assume
$g(−2) = g(−1) = 1$. The function `compute_terms` should have no loops and no
print statements. Further, for fast execution, this function should have only a single
recursive call.

```python
def compute_terms(n):
    if n==0:
        return (1,1,1) # Returns (g(0),g(-1),g(-2))
    s1,s2,s3=compute_terms(n-1) #s1=g(n-1) ; s2=g(n-2) ; s3=g(n-3) 
    g=s1+s2+s3-1 # g(n)=g(n-1)+g(n-2)+g(n-3)-1
    return (g,s1,s2) 
```

- Step 2: Write a program that does the following:
    - Ask the user to enter a non-negative integer $n$.
    - If $n < 0$, terminate the program.
    - If $n \ge 0$, compute the value of $g(n)$ by using the `compute_terms` with an appropriate parameter and print the value of $g(n)$.

```python
n=int(input("Enter a integer"))
if n<0:
    print("Terminating the program...")
    exit 
else:
    res=compute_terms(n)
    print(f"The required value of g({n}):{res[0]}")
```
