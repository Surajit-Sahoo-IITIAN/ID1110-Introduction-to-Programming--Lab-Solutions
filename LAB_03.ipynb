```python
##Q.6 (Printing first k prime numbers for a given k between 1 and 500 in the decreassing order)##

n = int(input("Enter an integer n between 1 and 500: "))
while n < 1 or n > 500:
    print("Out of range. Try again.")
    n = int(input("Enter an integer n between 1 and 500: "))
    continue
k = n
print(k)
count = 0
print(f"first {k} prime numbers are:")
# Your logic for starting the sequence
L=[]
if k == 1:
    L.append(2)
    count = 1
elif k >= 2:
    L.append(2)
    L.append(3)
    count = 2
    # if k > 2:
    #     print(", ", end="")
current=4
while count<k:
    is_prime=True
    for i in range(2,(current//2)+1):
        if current%i ==0:
            is_prime=False
            break
    if is_prime==True:
        L.append(current)
        count += 1
    
    current += 1 
print(L)
print(f"first {k} prime numbers in reverse order are:")
L1=[]
for i in range(len(L)):
    L1.append(L[len(L)-i-1])
print(L1)
```

```python
##Q.7. Grading system using if - else ##
"""Concept :: 1. Width specifies the total number of character spaces reserved for output.
              2. alignment determines whether the value is placed to the left, right, or center within that space. 
   Syntax:: {value:<width}   # left aligned
            {value:>width}   # right aligned
            {value:^width}   # center aligned
"""
n = int(input("Enter number of students: "))
marks_list = []
grades_list = []

for i in range(1, n + 1):
    mark = int(input(f"Enter the mark for student {i}: "))

    while mark < 0 or mark > 100:
        print(f"Enter the correct mark for student {i}")
        mark = int(input(f"Enter the mark for student {i}: "))

    marks_list.append(mark)

    if mark >= 90:
        grade = 'S'
    elif mark >= 75:
        grade = 'A'
    elif mark >= 65:
        grade = 'B'
    elif mark >= 55:
        grade = 'C'
    elif mark >= 45:
        grade = 'D'
    elif mark >= 35:
        grade = 'E'
    else:
        grade = 'F'

    grades_list.append(grade)

for i in range(n):
    print(f"Mark:{marks_list[i]:>3} Grade:{grades_list[i]:^3}")
```

```python
##Q.8. Grading system using Match - Case ##
## Concept :: The match–case statement matches the value of an expression against different cases and executes the block corresponding to the first matching case.##
n = int(input("Enter number of students: "))

marks_list = []
grades_list = []

for i in range(1, n + 1):
    mark = int(input(f"Enter marks of student {i}: "))

    while mark < 0 or mark > 100:
        print("Invalid mark! Enter 0–100 only.")
        mark = int(input(f"Enter marks of student {i}: "))

    marks_list.append(mark)

    match mark // 10:
        case 10 | 9:
            grade = 'S'
        case 8:
            grade = 'A'
        case 7:
            grade = 'B'
        case 6:
            grade = 'C'
        case 5:
            grade = 'D'
        case 4:
            grade = 'E'
        case _:
            grade = 'F'

    grades_list.append(grade)

print("\nMarks and Grades:")
for i in range(n):
    print(f"{marks_list[i]:>3} {grades_list[i]:^3}")
```

```python
## HW Q.2. Grading System with only one list ##

n = int(input("Enter number of students: "))
marks_list = []
for i in range(1, n + 1):
    mark = int(input(f"Enter marks of student {i}: "))

    while mark < 0 or mark > 100:
        print("Invalid mark! Enter an integer between 0 and 100.")
        mark = int(input(f"Enter marks of student {i}: "))
    marks_list.append(mark)

print("\nMarks and Grades:")
for i in range(n):
    mark = marks_list[i]

    if mark >= 90:
        grade = 'S'
    elif mark >= 75:
        grade = 'A'
    elif mark >= 65:
        grade = 'B'
    elif mark >= 55:
        grade = 'C'
    elif mark >= 45:
        grade = 'D'
    elif mark >= 35:
        grade = 'E'
    else:
        grade = 'F'

    print(f"{mark:>3} {grade:^3}")
```
