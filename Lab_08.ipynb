## Q.1. Anagrams Checking ##

Two words are said to be anagrams if they contain the same letters with the same
frequencies, possibly in a different order (for example, listen and silent ). Write a Python
program that determines whether two given words are anagrams using the following
procedure.
1. Ask the user to enter two words.
2. For each word, create a dictionary that stores each character as a key and its
frequency as the value.
3. Compare the two dictionaries.
4. Print True if the words are anagrams; otherwise print False.\
**Note: Don’t use counters.**

**Method1: Using Build-in functions** 
- `d.update({k,v})` ---> A method for adding key-value pair (k:v) to dictionary `d`.
- `s1.count(ch)` ---> frequency of character `ch` in a string `s1`.

```python
def Check_Anagrams(s1,s2):
    dict1={}
    dict2={}
    for ch in s1:
        dict1.update({ch:s1.count(ch)})
    for ch in s2:
        dict2.update({ch:s2.count(ch)})
    if dict1==dict2:
        return True
    else:
        return False
str1=input("Enter your first word:")
str2=input("Enter your second word:")
T=Check_Anagrams(str1,str2)
if T==True:
    print("Anagrams")
else:
    print("Not Anagrams")
```

**Method2: Using only `membership` operator**


```python
def Check_Anagrams(s1,s2):
    dict1={}
    dict2={}
    for ch in s1:
        if ch in dict1:
            dict1[ch]+=1
        else:
            dict1[ch]=1
    for ch in s2:
        if ch in dict2:
            dict2[ch]+=1
        else:
            dict2[ch]=1
    return dict1==dict2
str1=input("Enter your first word:")
str2=input("Enter your second word:")
T=Check_Anagrams(str1,str2)
if T==True:
    print("Anagrams")
else:
    print("Not Anagrams")
```

## Q.2. Classifying Numbers as Prime, Composite, Even and Odd ##

Recall that a prime number is a natural number that has exactly two factors.
- Step1: Define a function `factors(n)` that returns the list of all factors of $n$. For example,
factors(6) = $[1, 2, 3, 6]$.

```python
def factors(n):
    lst1=[]
    for i in range(1,n+1):
        if n%i==0:
            lst1.append(i)
    return lst1
```

- Step2: Using the `factors(n)` function, define another function `isprime(n)` that returns
`True` if $n$ is prime and `False` otherwise.

```python
def isprime(n):
    if len(factors(n))==2:
        return True
    else:
        return False
```

- Step3: Write a Python program that takes a positive integer $n \ge 2$ as input and does the following:
    - Create four dictionaries named `primes`, `composites`, `evens`, and `odds`. The
    names of the dictionaries are self-defining, i.e., they indicate what type of numbers
    they should contain. In each dictionary, the keys should be the corresponding
    integers, and the value associated with every key should be True.
    For example, in the dictionary primes, 2 should appear as a key but 6 should
    not.
    - For every integer $i$ from $2$ to $n-1$:
       - store $i$ in `primes` if it is prime, otherwise in `composites`,
       - store $i$ in `evens` if it is even, otherwise in `odds`.\
          Print all 4 dictionaries.

```python
primes={}
composites={}
evens={}
odds={}
n=int(input("Enter a positive integer greater than 2:"))
for i in range(2,n):
    if isprime(i):
        primes.update({i:isprime(i)})
    else:
        composites.update({i:not isprime(i)})
for i in range(2,n):
    if i%2==0:
        evens.update({i:"True"})
    else:
        odds.update({i:"True"})
print(f"Dictionary primes:{primes}")
print(f"Dictionary composites:{composites}")
print(f"Dictionary evens:{evens}")
print(f"Dictionary odds:{odds}")
```

- Step 2: Create a new dictionary called `avg_scores` where
     - the keys are the roll numbers, and
     - the values are 6-tuples of the form (m1,m2,m3,m4,m5, avg)
     where mi is the mark in course i and avg is the average of the marks.
     - If a student did not enroll in a particular course, store -1 for that course. The
    average should be computed only over the courses the student enrolled.
     - If student ’101’ enrolled only in courses 1, 2, and 4 and scored 72, 71, and 70,
    then the key–value pair should look like
    ’101’ : (72, 71,−1, 70,−1, 71.00)

```python
rolls= set(d1) | set(d2) | set(d3) | set(d4) | set(d5)
avg_scores={}
for r in rolls:
    m1=d1.get(r,-1)
    m2=d2.get(r,-1)
    m3=d3.get(r,-1)
    m4=d4.get(r,-1)
    m5=d5.get(r,-1)
    marks = [m1,m2,m3,m4,m5]
    valid_marks = []
    for m in marks:
        if m != -1:
            valid_marks.append(m)
    avg = sum(valid_marks)/len(valid_marks)
    avg_scores[r] = (m1,m2,m3,m4,m5,round(avg,2))
print(avg_scores)
```

## Homework : Memoization Vs Recursion ##

First we need to understand what is memoization and recursion technique for solving a problem.\
**Q. What is Memoization?**\
Ans: Memoization is a technique in which the results of previously computed function calls are stored (for example in a list or dictionary) so that if the same computation is needed again, the stored result can be reused instead of recomputing it.

**Q. What are differences between memoization and recursion?**\
Ans:

| Recursion | Memoization |
|-----------|-------------|
| 1. A function calls itself to solve a problem. | 1. Stores results of previous computations to avoid repeating them. |
| 2. May recompute the same values many times. | 2. Avoids repeated computations by reusing stored values. |
| 3. Can be inefficient for some problems (like Fibonacci). | 3. Improves efficiency by saving intermediate results. |

**Q. Which is more efficient?**\
Ans: Momoization is more efficient than recursion. The below code will show you the differences between running time of two techniques.

```python
import time
# Plain recursive Fibonacci
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n-1) + fib_recursive(n-2)
# Memoization (top-down) 
memo = {0: 0, 1: 1}
def fib_memo(n):
    if n in memo:
        return memo[n]
    memo[n] = fib_memo(n-1) + fib_memo(n-2)
    return memo[n]
n = 35
# Time recursive version
start = time.time()
r1 = fib_recursive(n)
t1 = time.time() - start
# Time memoized version
start = time.time()
r2 = fib_memo(n)
t2 = time.time() - start
result = {
    "n": n,
    "recursive_result": r1,
    "memoized_result": r2,
    "recursive_time_seconds": t1,
    "memoized_time_seconds": t2
}

result
```

**Conclusion:** For the above problem memoization technique takes less time than recursion.
