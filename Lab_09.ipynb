## Q.1. Set Operations ##

- Step 1 : Write a function `only_in_first_two` that takes three sets of integers $X$, $Y$ , and $Z$
as parameters and returns a set containing the elements that appear in $X$ and $Y$
but not in $Z$. There should not be any print statements in this function.\
**Recall the different set operations:**
- Union: `S1.union(S2)` or, `S1|S2`
- Intersection: `S1.intersection(S2)` or, `S1&S2`
- Difference: `S1.difference(S2)` or, `S1-S2`
- Symmetric Difference: `S1.symmetric_difference(S2)` or, `S1^S2`

```python
def only_in_first_two(X,Y,Z):
    Z1=X.intersection(Y)
    Z2=Z1.difference(Z)
    return Z2
```

- Step 2: Write a main program that takes elements of three sets $A$, $B$, and $C$ as input and,
using the function `only_in_first_two` with appropriate parameters, compute the
following and print the results:
    - the set of elements that appear in $A$ and $B$ but not $C$
    - the set of elements that appear in $B$ and $C$ but not $A$
    - the set of elements that appear in $A$ and $C$ but not $B$\
**Recall the `add` function:** `S1.add(elements)` ---> adding element `elements` to set `S1` 

```python
A=set()
n1=int(input("Enter the number of elements for set A"))
for i in range(n1):
    A.add(int(input("Enter elemets")))
B=set()
n2=int(input("Enter the number of elements for set B"))
for i in range(n2):
    B.add(int(input("Enter elemets")))
C=set()
n3=int(input("Enter the number of elements for set C"))
for i in range(n3):
    C.add(int(input("Enter elemets")))
print(A,B,C)
print(f"The set of elements that appear in A and B but not C : {only_in_first_two(A,B,C)}")
print(f"The set of elements that appear in B and C but not A : {only_in_first_two(B,C,A)}")
print(f"The set of elements that appear in A and C but not B : {only_in_first_two(A,C,B)}")
```

- Step 3: Using the above, in your main program, compute the set of elements that appear
in exactly two of the three sets A, B, and C and print it.

```python
exactly_two = only_in_first_two(A,B,C) | only_in_first_two(B,C,A) | only_in_first_two(A,C,B)
print("Elements appearing in exactly two sets:", exactly_two)
```

## Q.2. Sentence Reversal ##

- A sentence is a string that may contain spaces, and a word of this sentence is a maximal
sequence of consecutive letters without a space. Write a function `word_reversal` that
takes a sentence $S$ as a parameter and returns another string in which the order of words
is the opposite of that in $S$. There should be no print statements inside the function.
The main program asks the user for a sentence as input and prints another sentence
where the order of the words is reversed.
- Example:
‘My name is Pratik’ → ‘Pratik is name My’

- Using Built-in function `split()`

```python
def word_reversal(S):
    L=S.split()
    rev=""
    for i in range(len(L)):
     rev+=L[len(L)-i-1]
     if i != len(S):
             rev += " "
    return rev
```

- Without using built-in function `split()`
  

```python
def word_reversal(S):
    all_words=[]
    sin_word=""
    for ch in S:
        if ch!=' ':
            sin_word+=ch
        else:
            if sin_word!="":
                all_words.append(sin_word)
                sin_word=""
    if sin_word!="":
        all_words.append(sin_word)
    rev=""
    for i in range(len(all_words)):
        rev+=all_words[len(all_words)-i-1]
        if i != len(S):
            rev += " " 
    return rev
```

```python
S="My name is Pratik"
print(f"Reversed sentence:{word_reversal(S)}")
    
```

## Q.3. String partition ##

- Step 1: Write a function `split_into_groups` that takes a string $X$ as a parameter and returns
a list $L_x$ of strings such that:
     - If a letter appears twice or more `consecutively` in $X$, the letters must appear as a
   single element in the list $L_x$.
     - Otherwise, the letter should appear as a singleton string.
     - Example:
        - `'committee'` → `['c','o','mm','i','tt','ee']`
        - `'aaabbc'` → `['aaa','bb','c']`

```python
def split_into_groups(X):
    if X==" ":
        return []
    Lx=[]
    consecutive_group=X[0]
    for i in range(1,len(X)):
        if X[i]==X[i-1]:
            consecutive_group+=X[i]
        else:
            Lx.append(consecutive_group)
            consecutive_group=X[i]
    Lx.append(consecutive_group)
    return Lx
    
```

- Step 2: In the main program, we take a string as input and print the list of strings as defined
above.

```python
X=input("Enter your string")
print(f"String partition for the string {X} : {split_into_groups(X)}")
```

## Q.4. Most Frequent and Least Frequent elements in a Tuple ##

Write a function `most_freq_least freq` that takes a tuple $T$ as a input and return the a
tuple $(x,y)$ where $x$ is the most frequent element of $T$ and $y$ is the least frequent element
of $T$. Use a counter to solve the problem.
The main function should take a tuple as input, call the function `most_freq_least_freq`,
and print a tuple containing the most and least frequent elements.\


```python
from collections import Counter
def most_freq_least_freq(T):
    dic=Counter(T)
    most_freq=-1
    least_freq=float('inf')
    most_key=None
    least_key=None
    for k in dic:
        if dic[k]>most_freq:
            most_freq=dic[k]
            most_key=k
        if dic[k]<least_freq:
            least_freq=dic[k]
            least_key=k
    return (most_key,least_key)
            
```

**Concept used for Taking multiple elements in a single line of code**
- Using `map()`:
  - `map()` is a built-in Python function that takes a `function` and one or more iterables as input and returns an iterator(map object) that applies the given function to each element of the iterable(s).
  - Syntax: `map(function,iterable1,iterable2,....)`
            

```python
T = tuple(map(int, input("Enter elements of tuple (comma separated): ").split(",")))
result = most_freq_least_freq(T)
print("Most and least frequent elements:", result)
```

- Using List Comprehension:
    - List comprehension is a concise way to create a list by iterating over an iterable and optionally applying a condition.
    - Syntax:`[expression for x in iterable if condition]`

```python
n=int(input("Enter the number of tuple elements you want:"))
T=[int(x) for x in input("Enter Elements:").split(",")] 
T1=tuple(T)
result = most_freq_least_freq(T1)
print("Most and least frequent elements:", result)
```
