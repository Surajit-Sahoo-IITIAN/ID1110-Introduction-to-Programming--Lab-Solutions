## Q.1. Recursive implementation of factorial function ##

- Step1: Write a recursive `factorial` function that takes an integer $n$ as parameters and returns 1, if
$n < 1$ and returns the value of n factorial otherwise. For the second case, use a recursive call with
parameter $n−1$; a temporary variable temp to store the output of the recursive call and then return
$n * \text{temp}$. 

```python
def factorial(n):
    if n<1:
        return 1
    else: 
        temp=factorial(n-1)
        return n*temp
```

- Step2:  Write a program that takes an integer $x$ as input from user, invokes the factorial function and
prints the computed value.

```python
x=int(input("Enter an integer:"))
com_val=factorial(x)
print(f"The value of the {x}! = {com_val}")
```

## Q.2. Recursive implementation of binary search ##

- Step1: Write a function `input_ascending_sequence` which takes an integer $n$ as parameter, and does the
following: Initialize a list $l$ to be empty. If $n > 0$, ask the user to give $n$ integers in non-decreasing
order as input. After taking each of the $n$ numbers, if is less than the previous value, then ignore
it and prompt the user to enter a valid input. If the input is valid, append it to $l$. Finally return
the list.

```python
def input_ascending_sequence(n):
    l=[]
    if n>0: 
        print(f"Enter {n} integers in non-decreasing order:")
    while len(l)<n:
        new=int(input(f"Enter {len(l)+1} element:"))
        if len(l)==0 or new>=l[len(l)-1]:
            l.append(new)
        else:
            print("Enter a valid input")
    return l
```

- Step2: Write a recursive function `bin_search` which takes a list $l$, a number $x$ and two indices first and
last as parameters and assuming $l$ is a non-decreasing list, does a binary search for $x$ in the sublist
of $l$ between positions first to last (both inclusive). The function should return $−1$ if $x$ is not
present in the sublist and otherwise return an index $i$ with $\text{first} ≤ i ≤ \text{last}$ with $l[i] = x$.

```python
def bin_search(l,x,first,last):
    if first>last:
        return -1
    mid=(first+last)//2
    if l[mid]==x:
        return mid
    elif x<l[mid]:
        return bin_search(l,x,first,mid-1)
    else:
        return bin_search(l,x,mid+1,last)
    
```

- Step 3: Write a program that does the following:
    - Ask the user to enter a positive integer $n$. If $n <= 0$, terminate the program.
    - If $n > 0$, invoke the function `input_ascending_sequence` with $n$ as parameter and store the
    result in a list **inp_list**.
    - Ask the user to enter an integer key $k$ to search in the list **inp_list**.
    - Invoke `bin_search` function to search for $k$ in the list **inp_list**.
    - If $k$ is found, print the index of $k$ in **inp_list**. Otherwise, print $k$ is not present in the list.

```python
n=int(input("Enter a positive integer"))
if n<=0:
    print("Program terminated!!")
else:
    inp_list=input_ascending_sequence(n)
    k=int(input("Enter the integer key k to search in the list:"))
    find=bin_search(inp_list,k,0,len(inp_list)-1)
    if find==-1:
        print(f"{k} is not present in the list")
    else:
        print(f"{k} is present in the list at index {find}")
```

## Q.3. Selection sort algorithm ##

- Step1: Write a `find_max_position` function that takes a list $l$ and two non-negative integers first and
last as parameters and returns the smallest index $k$ such that $\text{first} ≤ k ≤ \text{last}$ and a maximum
element in the sublist of $l$ from index first to index last (both inclusive) is present at index $k$. It
may be assumed that $0 ≤ \text{first} ≤ \text{last} <\text{length of}\ l$

```python
def find_max_position(l,first,last):
    max_pos=first
    for i in range(first+1,last+1):
        if l[i]>l[max_pos]:
            max_pos=i
    return max_pos
```

- Step2: Write a `swap_list_elements` function that takes a list $l$ and two indices $i$ and j as arguments and
modifies the list $l$ by swapping elements in positions $i$ and $j$. This function will not return any
value. It may be assumed that $0 ≤ i, j <\text{length of}\ l$.

```python
def swap_list_elements(l,i,j):
    l[i],l[j]=l[j],l[i]
```

- Step3:
    - Take an integer $n > 1$, followed by $n$ numbers which are elements of the list $l$ as input
    - For each index $k$ from $n-1$ downto $1$ (both inclusive):
        - Invoke `find_max_position` function to compute the index of the maximum element between
          positions $0$ and $k$ (both inclusive) in $l$. Let the position be **posn**.
        - Invoke `swap_list_elements` function to exchange elements in positions **posn** and $k$ in the
         list $l$.
        - Print the list $l$

```python
n=int(input("Enter a integer n>1"))
if n<=1:
    print("Invalid input!!")
else:
    l=[]
    for i in range(n):
        num=int(input(f"Enter the {i+1} element"))
        l.append(num)
    for k in range(n-1,0,-1):
        posn=find_max_position(l,0,k)
        swap_list_elements(l,posn,k)
    print(f"The sorted list:{l}")
```

## Q.4. Selection Sort algorithm with modification ##

- Step1: Same as **Step1** and **Step2** of **Q.3.**

- Step2:
   - Write a `sort_list` function that takes a list $l$ and a non-negative number $n$ as input and assuming
     $n ≤\text{length of}\ l$, does the following: If $n <= 1$, return without doing anything. If $n > 1$, modify $l$ so
     that the sublist of $l$ from position $0$ to position $n − 1$ gets sorted using the following method:
     - For each index $k$ from $n-1$ downto $1$ (both inclusive):
        - Invoke `find_max_position` function to compute the **index** of the maximum element between
        positions $0 $ and $k$ (both inclusive) in $l$. Let the position be **posn**.
        - Invoke `swap_list_elements` function to exchange elements in positions **posn** and $k$ in the
        list $l$.

```python
def sort_list(l,n):
    if n<=1:
        return 
    for k in range(n-1,0,-1):
        posn=find_max_position(l,0,k)
        swap_list_elements(l,posn,k)
```

- Step3:

## Q.5. Recursive implementation of selection sort algorithm ##

- Step1: Same as **Step1** and **Step2** of **Q.3.**

- Step2: Write a `sort_list_recursive` function that takes a list $l$ and a non-negative number $n$ as input
and assuming $n ≤\text{length of}\ l$, does the following: If $n <= 1$, return without doing anything. If
$n > 1$, modify $l$ so that the sublist of $l$ from position $0$ to position $n − 1$ (both inclusive) gets sorted
using the following method:
    - Invoke `find_max_position` function to compute the **index** of the maximum element between
    positions $0$ and $n − 1$ (both inclusive) in $l$. Let the position be **posn**.
    - Invoke `swap_list_elements` function to exchange elements in positions **posn** and $n − 1$ in the
    list $l$.
    - Recursively invoke `sort_list_recursive` function to sort the sublist of $l$ from position $0$ to
    position $n − 2$ (both inclusive).

```python
def sort_list_recursive(l,n):
    if n<=1:
        return 
    else:
        posn=find_max_position(l,0,n-1)
        swap_list_elements(l,posn,n-1)
        sort_list_recursive(l,n-1)
```

- Step3:
    - Take an integer $n > 1$, followed by $n$ numbers which are elements of the list $l$ as input
    - Invoke `sort_list_recursive` function to sort the list $l$ from position $0$ to position $n − 1$ (both
    inclusive).
    - Print the final list.

```python
n = int(input("Enter n (>1): "))
if n <= 1:
    print("n must be greater than 1.")
else:
    l = []
    print(f"Enter {n} elements:")
    for i in range(n):
        num=int(input(f"Enter the {i+1} element"))
        l.append(num)
    sort_list_recursive(l, n)
    print("Final sorted list:", l)
```
