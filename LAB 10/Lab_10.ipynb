## Q.1 CSV-Based Student Marks Aggregation System ##

You are given a directory named Marklists containing some CSV (Comma Separated
Values) files, each containing mark list for one course (number of courses is not known
in advance). Each file contains data in the following two column format:
<div style="text-align:center;">
    
RollNo, Marks  
1001, 33  
1005, 22  
...

</div>


 **Step:** Your task is to create a new CSV file which contains the combined marklists of all these
courses. Use the method given below.
- Write a function `markfile_to_dictionary` which takes as parameter the filename
of one marklist in the above format and returns dictionary in which the key is the
roll number and value is the marks obtained for that course. For this,
     - Open the file, if it exists and read the first line of the file and check if it is in
proper format. If the file does not exist or is not in proper format, return an
empty dictionary. Otherwise, proceed as below.
     -  Read each subsequent line in the file, strip the special characters (like ‘\n’
and spaces) from the beginning and end; and split into rollno and marks.
If the roll number is already present as a key in the dictionary, skip the line.
Otherwise, insert the data into the dictionary.
    - After processing all lines, return the dictionary created. Make sure that the file
object is closed before you return from the function.

```python
## List of files inside marklist
import os
print(os.listdir("marklist"))
```

```python
## Check the header for each csv_file 
csv_files=["marklist/course_1.csv","marklist/course_2.csv","marklist/course_3.csv"]
for filename in csv_files:
    with open(filename,"r") as f:
        print(f"Header for {filename}: {f.readline().strip()}")
```

```python
def markfile_to_dictionary(filename):
    roll_mark={}
    try:
        with open(filename,"r") as f:
            header=f.readline().strip()
            if header not in ["roll,course_1.txt","roll,course_2.txt","roll,course_3.txt"]:
                return {}
            for line in f:
                lines=line.strip()
                if lines=="":
                    continue
                two_parts=lines.split(",")
                if len(two_parts)!=2:
                    continue
                roll=two_parts[0].strip()
                marks=two_parts[1].strip()
                if roll in roll_mark:
                    continue
                roll_mark[roll]=int(marks)
        return roll_mark
    except FileNotFoundError:
        return {}        
```

**Step-2:** In your main program,
- Find out the list of all CSV files in the Marklists directory and by invoking
the markfile to dictionary function, convert them to dictionaries.

```python
import os
folder = "marklist"
files = [f for f in os.listdir(folder) if f.endswith(".csv")]
all_dicts=[]
for file in files:
    filepath=os.path.join(folder,file)
    d=markfile_to_dictionary(filepath)
    if d:
        all_dicts.append(d)
d1=all_dicts[0]
d2=all_dicts[1]
d3=all_dicts[2]
print("Dictionary 1:")
print(d1)
print("Dictionary 2:")
print(d2)
print("Dictionary 3:")
print(d3)
```

- Count the number valid marklists. Let it be k (which is not known in advance).

```python
k=len(all_dicts)
```

```python
## Another way to get all roll numbers
# all_rolls=set(d1)|set(d2)|set(d3)
```

- Create a new dictionary `Combined_Marks` in which the key is the roll number
and value is a list containing k marks (one per course, with mark -1 if student
is not crediting the course), followed by the average of non-negative scores.

```python
Combined_Marks={}
all_rolls=set()
for d in all_dicts:
    all_rolls.update(d.keys())
print(all_rolls)
for roll in all_rolls:
    List_marks=[]
    for d in all_dicts:
        if roll in d:
            List_marks.append(int(d[roll]))
        else:
            List_marks.append(-1)
    valid_marks=[m for m in List_marks if m!=-1]
    if valid_marks:
        avg=sum(valid_marks)/len(valid_marks)
    else:
        avg=0
    List_marks.append(round(avg,2))
    Combined_Marks[roll]=List_marks
print(Combined_Marks)
```

- Write the contents of the dictionary Combined Marks to a file `score_summary.csv` in the following format:\
RollNo,C1,C2,C3,C4,...,Ck,Average\
1001,22,32,-1,40,.....,10,26\
............................

```python
with open("score_summary.csv","w") as f:
    header=["RollNo"]
    for i in range(k):
        header.append(f"C{i+1}")
    header.append("Average")
    print(header)
    for i in range(len(header)):
        if i==0:
          f.write(header[i])
        else:
            f.write("," + header[i])
    f.write("\n")
    for roll in sorted(Combined_Marks.keys()):
        marks_avg = Combined_Marks[roll]
        f.write(roll)
        for i in range(len(marks_avg)):
            f.write("," + str(marks_avg[i]))
        f.write("\n")     
```

## Q.2. Caesar Cipher ##

The `Caesar cipher` is an encoding scheme in which for each character $x$ in the input
plaintext is replaced by another character $y$, using a certain mapping. The new text
obtained is called the `Cipher text`. If we know the encoding scheme and the Cipher
text, the original text can be recovered.
In our problem, the mapping of letters is done using the following formula:
<center>
<ht>$f(i) = (5i + 3) \ mod \ 26$ </ht>
    </center>
where $i$ is the position of the letter ($i = 0$ for $a$ and $A$, $i = 1$ for $b$ and $B$, and so on),
with uppercase letters mapped to uppercase letters and lowercase letters to lowercase
letters. Other characters map to themselves.

- Step 1: Write a function `secret_encode` that takes a string as parameter and returns the
corresponding Cipher text, using the encoding formula given above.

```python
def secret_encode(plaintext):
    cipher_text=""
    for ch in plaintext:
        if ch.isalpha():
            if ch.islower():
                plain_position=ord(ch)-97
                i=plain_position
                cipher_position=(5*i+3)% 26
                cipher_text=cipher_text+chr(cipher_position+97)
            else:
                plain_position=ord(ch)-65
                i=plain_position
                cipher_position=(5*i+3)% 26
                cipher_text=cipher_text+chr(cipher_position+65)
        else:
            cipher_text+=ch
    return cipher_text
        
```

- Step 2: In your main program, read the contents of a file `input.txt` and encode it using
the `secret_encode` function and write the Cipher text obtained into another file
`secret.txt`.

```python
import os
with open("input.txt","r", encoding="utf-8") as f:
    content=f.read()
with open("secret.txt","w", encoding="utf-8") as f:
    cipher_text=secret_encode(content)
    f.write(cipher_text)
```
