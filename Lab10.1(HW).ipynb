## Q.1. List Comprehenssion ##

- Write a program that takes a positive number $n$, followed by
elements of a list $l$ containing $n$ integers and prints each of the following information
(using list comprehension syntax for computations) :
- List $l_1$ which contains elements of $l$ which are multiples of $5$, followed by average of
elements of $l_1$, if $l_1$ is non empty
- List $l_2$ which contains elements of $l$ which are greater than or equal to zero, followed
by average of elements of $l_2$, if $l_2$ is non empty
- List $l_3$ which contains square of every element of $l$ which is a multiple of 5.
- List $l_4$ which contains all triples of elements which are Pythagorean triples: That
is all triples $(x, y, z)$ of $l$ with $x^2 + y^2 = z^2$ and $0 < x < y < z$. The condition
$x < y < z$ ensures that if $(x, y, z)$ is listed, then $(y, x, z)$ is not listed.

```python
def list_compre(l):
    print(f"The original list l={l}")
    if len(l)==0:
        raise ValueError("Input list should not be empty")
    l1=[m for m in l if m%5==0]
    if len(l1)!=0:
        l1.append(sum(l1)/len(l1))
    else:
        print("l1 is empty no avg to compute")
    l2=[m for m in l if m>=0]
    if len(l2)!=0:
        l2.append(sum(l2)/len(l2))
    else:
        print("l2 is empty no avg to compute") 
    l3=[m*m for m in l if m%5==0]
    if len(l3)!=0:
        l3.append(sum(l3)/len(l3))
    else:
        print("l3 is empty no avg to compute")  
    l4=[(x,y,z) for x in l for y in l for z in l if x**2+y**2==z**2 and 0<x<y<z]
    print(f"The list l1:{l1}")
    print(f"The list l2:{l2}")
    print(f"The list l3:{l3}")
    print(f"The list l4:{l4}")
    
```

## Q.2. Flattend List ##

A flattened list is a list obtained from a nested list by recursively extracting all non-list
elements and arranging them into a single, one-dimensional list while preserving their
original order. Write a function `flatten_list(L)` that takes a nested list as an input
and returns a flattened list containing all the integers.

The input list may contain integers or other nested lists to arbitrary depth. The solution
must use recursion. You should not use any library function to flatten. For each element
of the list, we check whether it is an integer or another list. If it is an integer, we include
it directly in the output list. If it is a list, we recursively apply the same function to
that sublist to extract its elements.\
For example,
`Input: [1, [2, [3, 4], 5], 6]`\
`Output: [1, 2, 3, 4, 5, 6]`

```python
def flatten_list(L):
    output_list = []
    for element in L:
        if isinstance(element, int):      
            output_list.append(element)
        elif isinstance(element, list):   
            output_list+=flatten_list(element)
    return output_list
```

```python
l=[1,[2,[3,4],5],6]
print("Input list:",l)
print("Flatten list:",flatten_list(l))
```

## Q.3. Isogram String ##

Write a function `is_isogram(S)` that takes a string `S` as input and returns `True` if every
letter appears at most once in S, otherwise returns `False`.


- First, create a list where each element is a character of the string. Then, using list
comprehension, create another list containing characters that appear exactly once in the
original list. Finally, compare the two lists to determine whether the string is an isogram
or not.
    - Input: "lamp"
    - Output: True
    - Input: "letter"
    - Output: False

```python
def is_isogram(S):
    l=[]
    for char in S:
        l.append(char)
    exact_once=[x for x in l if l.count(x)==1]
    if l==exact_once:
        return True
    else:
        return False
```

```python
print(is_isogram("lamp"))
print(is_isogram("letter"))
```

## Q.4. Isomorphism of Two String ##

- Two strings $S_1$ and $S_2$ are isomorphic if each character $\alpha$ in $S_1$ can be mapped to exactly
one character $f(\alpha)$ in $S_2$, keeping the relative positions of $\alpha$ in $S_1$ and $f(\alpha)$ in $S_2$ the
same and such that no two characters map to the same character. Write a function
`are_isomorphic(s, t)` that takes two strings as input and returns `True` if the strings
are isomorphic, otherwise returns `False`.

In short the following two rules must be implemented:
- Rule 1: Same character in $S_1$ must always map to the same character in $S_2$
- Rule 2: Two different character in $S_1$ cannot map to the same character in $S_2$

```python
def are_isomorphic(S1,S2):
    if len(S1)!=len(S2):
        return False
    s_t={}
    t_s={}
    for i in range(len(S1)):
        first_S1=S1[i]
        first_S2=S2[i]
        if first_S1 in s_t:
            if s_t[first_S1]!=first_S2:
                return False
        else:
            s_t[first_S1]=first_S2
        if first_S2 in t_s:
            if t_s[first_S2]!=first_S1:
                return False
        else:
            t_s[first_S2]=first_S1
    return True     
```

```python
print(are_isomorphic("egg", "add"))    
print(are_isomorphic("foo", "bar")) 
```
