## Q.1. Tower Of Hanoi (TOH) Problem ##

- Step 1: Write a recursive function `Towers` that takes the number of disks $n$, source, target and mid as a
parameters (assuming $n \ge 1 $and the values of source, target and mid are distinct), prints the
sequence of moves of a valid Tower of Hanoi game with $n$ disks, with all disks in the correct order
intitially on source pole and at the end of the game in correct order on target pole. The function
should return the number of moves. 


```python
def Towers(n,source,target,mid):
    if n==1:
        print(f"Disk {n} :{source} ---> {target} ")
        return 1
    else:
        count1=Towers(n-1,source,mid,target)
        print(f"Disk {n} :{source} ---> {target} ")
        count2=Towers(n-1,mid,target,source)
        return 1+count1+count2
```

- Step 2: Write a main program which asks the user to enter number of disks $n$, invokes the `Towers` function
with parameters $n, A, B, C$. The function will print the sequence of moves. After these, in the main
program, print the total number of moves.

```python
n=int(input("Enter the number of disks:"))
print("..........Sequence of moves...........")
Total_moves=Towers(n,'A','B','C')
print(f"Tower number of moves for TOH problem with {n} disks is/are {Total_moves}")
```

- Step 1: Write a function `remove_duplicates` that takes a list $l$ as input and returns another list after
removal of duplicates from $l$. This function should not have any print statements.

- Approach 1: Sorted the list and compare the consecutive elements. But the returned list in this case **does not preserved** the order of original list elemets. 

```python
# def remove_duplicates(l):
#     if len(l) == 0:
#         return []
#     l = sorted(l)          
#     result = [l[0]]
#     for i in range(1, len(l)):
#         if l[i] != l[i - 1]:
#             result.append(l[i])
#     return result
```

- Approach 2(Recommended): Use membership operater to find duplicates. The returned list **Preserved** the original list elemets order.

```python
def remove_duplicates(l):
    no_duplicates=[]
    for x in l:
        if x not in no_duplicates:
            no_duplicates.append(x)
    return no_duplicates
```

- Step 2:  Write a main program that takes an integer $n > 1$, followed by $n$ numbers and does the following:
    - Store the $n$ numbers into a list $l$.
    - Invoke `remove_duplicates` function with $l$ as parameter.
    - Print the elements of the list returned by the function.

```python
n=int(input("Enter the number of elemets of the list:"))
L=[]
print(f"Enter {n} elements of the list")
for i in range(n):
    L.append(int(input()))
print(f"Original List(may with duplicates): {L}")
No_Dup=remove_duplicates(L)
print(f"List after removing duplicates: {No_Dup}")
```

## Q.3. GCD using recursion ##

- Idea: The greatest common divisor (gcd) of two positive integers $a$ and $b$ is the largest integer $d$ such that $a$
and $b$ are both integer multiples of $d$. Obviously, $\gcd(a, b) = \gcd(b, a)$ and if $a$ is a multiple of $b$, then
$\gcd(a, b) = b$. Further, it is a well known fact that if $a > b$, then $\gcd(a, b) = \gcd(a\ mod\ b, b)$ (You may try
to prove it yourself). The famous `Euclid’s algorithm` is a recursive algorithm to find gcd of two numbers,
using these facts.

**Claim:** $\gcd(a,b)=\gcd(b,a\ mod\ b)$ \
**Proof:** Let $d=\gcd(a,b) \ \& d'=\gcd(a \ mod \ b,b)$ \
$\gcd(a,b)=d \implies d|a \ \& \ d|b \implies d|(a-bq)=a \ mod \ b$, for some integer $b$ \
$d|a \ mod \ b \ \& \ d|a \implies d|d' (\because d'=\gcd(a \ mod \ b,a))$ \
$d'=\gcd(a \ mod \ b,b) \implies d'| a \ mod \ b= (a-bq) \ \& \ d'|b \implies d'|(a-bq)+bq=a $ \
$d'|a \ \& d'|b \implies d'|\gcd(a,b)=d$ \
$\because d|d' \ \& \ d'|d \ \& \ (d,d')>(0,0) \implies \boxed{d=d'}$


- Step 1: Write a recursive function `gcd_recursive` that takes two positive intgers $a$ and $b$ and assuming $a \ge b$, returns the $\gcd$ of $a$ and $b$, using the facts given.
    - The base case of your function is when $a$ is a multiple of $b$.
    - In other cases, you have to recursively call the same function with appropriate parameters.
    - Note that in the recursive call your first parameter should be always greater than or equal to the second parameter.
    - To ensure termination of the recursion, you may ensure that the value of the first parameter is decreasing in each recursive call. This function should not have any print statements.

```python
def gcd_recursive(a,b):
    if a%b==0:
        return b
    else:
        return gcd_recursive(b,a%b)
```

- Step 2: Write a program that takes two positive integers $a$, $b$ as parameters, invoke the function `gcd_recursive`
to find their $\gcd$ and finally print the answer. \
**$\boxed{Remark}$**
\begin{aligned}
\gcd(a, 0) &= a, && a > 0 \\
\gcd(0, b) &= b, && b > 0 \\
\gcd(0, 0) &\text{ is undefined.}
\end{aligned}


```python
a,b=map(int,input("Enter two integers a and b separated by comma").split(","))
if a==0 and b==0:
    print(f"GCD({a},{b}) is undefined")
    print("Enter valid values of a and b and run the program again")
if b == 0 and a>0:
    print(f"GCD({a}, {b}) = {a}")
if a == 0 and b>0:
    print(f"GCD({a}, {b}) = {b}")
if a>=b and a!=0 and b!=0:
    print(f"GCD({a},{b})={gcd_recursive(a,b)}")
if a<b and b!=0 and a!=0:
    print(f"GCD({a},{b})={gcd_recursive(b,a)}")
```

## Q.4. Relative Primes in a List ##

- Step 1:  Two positive integers are said to be relatively prime if their gcd is $1$. Write a program that computes
all pairs of relatively prime elements from a given list, as desribed below:
    - Asks the user to enter a positive integer $n$.
    - If $n < 0$, terminate the program.
    - Otherwise, ask the user to enter $n$ positive integers. Do a proper validation during the input and
    store the $n$ numbers into a list $l$.
    - Use the `remove_duplicates` function defined earlier, to remove duplicates from $l$, and store the
    result to a new list $l_1$.
    - By using the `gcd_recursive` function defined in the previous question for each (unordered) pair of
    distinct elements from $l_1$, print each such pair which is relatively prime.

```python
n = int(input("Enter a positive integer n: "))
if n < 0:
    print("Invalid input. Program terminated.")
    exit()
l = []
count = 0
while count < n:
    x = int(input(f"Enter positive integer {count+1}: "))
    if x > 0:
        l.append(x)
        count += 1
    else:
        print("Please enter a positive integer.")
l_1 = remove_duplicates(l)
print("Relatively prime pairs:")
for i in range(len(l_1)):
    for j in range(i + 1, len(l_1)):
        a, b = l_1[i], l_1[j]
        if a >= b:
            g = gcd_recursive(a, b)
        else:
            g = gcd_recursive(b, a)
        if g == 1:
            print((a, b))
```

## HW: Recursive implementation of selection sort (Alternative way) ##

- Step 1: This exercise asks you to do an alternate recursive implementation of selection sort algorithm, for sorting
a list of integers in non-decreasing (i.e ascending) order. The only differences from Q5 of the previous
lab are the following:
    - Instead of `find_max_position` function, you should define a `find_min_position` function. The
    parameters of this function are the same as that of `find_max_position` defined last week. But, the
    return value should be the index of the minimum element, if the sublist of $l$ from first to last (both
    inclusive) is non-empty.
    - Write a `new_sort_list` recursive function that uses `find_min_position` with appropriate parameters and a recursive call to `new_sort_list` recursive for doing the selection sort.


```python
def find_min_position(l, first, last):
    min_pos = first
    for i in range(first + 1, last + 1):
        if l[i] < l[min_pos]:
            min_pos = i
    return min_pos

def swap_list_elements(l, i, j):
    l[i], l[j] = l[j], l[i]

def sort_list_recursive(l, first, last):
    if first >= last:
        return
    else:
        posn = find_min_position(l, first, last)
        swap_list_elements(l, posn, first)
        sort_list_recursive(l, first + 1, last)

n = int(input("Enter n (>1): "))
if n <= 1:
    print("n must be greater than 1.")
else:
    l = []
    print(f"Enter {n} elements:")
    for i in range(n):
        num = int(input(f"Enter the {i+1} element: "))
        l.append(num)
    sort_list_recursive(l, 0, n - 1)
    print("Final sorted list:", l)
```
